## Sección 5 — Listas y arreglos: varios valores a la vez

Una **lista** guarda varios valores en orden, entre corchetes `[ ]`.

````python
notas = [85, 90, 78, 92, 88, 76, 95]
estudiantes = ["Ana", "Bruno", "Carla", "Diego", "Elena", "Felipe", "Gloria"]

print(notas)
print(len(notas))   # ¿cuántos elementos tiene?
````

### Python empieza a contar desde 0

Esto sorprende a casi todos: el **primer** elemento tiene posición **0**.

| Posición (índice) | 0 | 1 | 2 | 3 | 4 | 5 | 6 |
|---|---|---|---|---|---|---|---|
| Valor | 85 | 90 | 78 | 92 | 88 | 76 | 95 |

Y al "cortar" un tramo con `[inicio:fin]`, el `fin` **no se incluye**.

````python
print(notas[0])      # primera nota
print(notas[2])      # tercera nota
print(notas[-1])     # última nota (índice negativo cuenta desde el final)
print(notas[1:4])    # posiciones 1, 2 y 3 (la 4 NO se incluye)
````

### El problema de las listas: no hacen matemáticas "por elemento"

Probemos sumar 2 puntos a cada nota:

````python
try:
    notas + 2
except TypeError as error:
    print("Error:", error)
````

Una lista no sabe qué hacer con `+ 2`. La solución es convertirla en un **arreglo de numpy** (`np.array`), que sí aplica las operaciones a **todos los elementos a la vez** (a esto se le llama *vectorización*):

````python
notas = np.array([85, 90, 78, 92, 88, 76, 95])

print(notas + 2)              # suma 2 a cada nota
print(notas * 0.1)            # el 10 % de cada nota
print(np.round(notas / 10, 1))  # divide todo y redondea a 1 decimal
````

````python
# Comparaciones: producen un arreglo de True/False
aprobo = notas >= 80
print(aprobo)

# Y podemos usarlo para filtrar (quedarnos solo con las notas que cumplen)
print(notas[aprobo])
````

### ✏️ Tu turno — Arreglos

## Sección 6 — Estadística básica sobre un arreglo

````python
print("Media:        ", np.mean(notas))
print("Mediana:      ", np.median(notas))
print("Desv. estándar:", np.std(notas, ddof=1))   # ddof=1 → desviación MUESTRAL (divide entre n-1)
print("Varianza:     ", np.var(notas, ddof=1))
print("Mínimo:       ", np.min(notas))
print("Máximo:       ", np.max(notas))
print("Suma total:   ", np.sum(notas))
````

> ⚠️ **Trampa frecuente:** `np.std()` y `np.var()` dividen entre *n* por defecto (desviación **poblacional**). Para la desviación **muestral** (la que casi siempre queremos, y la que usan Excel, R y pandas) hay que agregar `ddof=1`.

Para un resumen completo en una sola línea, convertimos el arreglo en una **Serie de pandas** y usamos `.describe()`:

````python
pd.Series(notas).describe()   # conteo, media, desv. estándar, mínimo, cuartiles y máximo
````

### Acceder a columnas, filas y celdas

- Una columna: `df["columna"]`
- Varias columnas: `df[["col1", "col2"]]` (¡doble corchete!)
- Por **etiqueta**: `df.loc[fila, columna]`
- Por **posición**: `df.iloc[fila, columna]`

## Sección 8 — Funciones: bloques de construcción

Una **función** recibe entradas (*argumentos*), hace algo con ellas y devuelve un resultado. Ya usaste muchas (`print`, `len`, `np.mean`…). También puedes **crear las tuyas**, algo clave para mantener tu código ordenado y no repetirlo.

Se definen con `def`, y **la sangría (los 4 espacios al inicio de cada línea) es obligatoria**: indica qué líneas pertenecen a la función.

````python
def celsius_a_fahrenheit(celsius):
    fahrenheit = (celsius * 9 / 5) + 32
    return fahrenheit

print(celsius_a_fahrenheit(0))     # 32
print(celsius_a_fahrenheit(100))   # 212
print(celsius_a_fahrenheit(37))    # temperatura corporal ≈ 98.6
````

Las funciones pueden tener varios argumentos y **valores por defecto**:

````python
def saludar_estudiante(nombre, idioma="ES"):
    if idioma == "ES":
        print("Bienvenido/a,", nombre, "— ¡Mucho éxito en el curso!")
    else:
        print("Welcome,", nombre, "— Good luck in the course!")

saludar_estudiante("Ana")
saludar_estudiante("Bruno", idioma="EN")
````

## Sección 10 — Un vistazo al flujo de modelado predictivo

Todo proyecto de predicción analítica sigue este flujo general. Construiremos cada paso en detalle a lo largo del curso.

```
 ┌──────────────┐     ┌──────────────┐     ┌───────────────┐
 │ 1. RECOLECTAR│────▶│ 2. LIMPIAR Y │────▶│ 3. EXPLORAR   │
 │    DATOS     │     │    PREPARAR  │     │    (EDA)      │
 └──────────────┘     └──────────────┘     └───────┬───────┘
                                                   │
 ┌──────────────┐     ┌──────────────┐     ┌───────▼───────┐
 │ 6. DESPLEGAR │◀────│ 5. EVALUAR   │◀────│ 4. CONSTRUIR  │
 │ Y MONITOREAR │     │    Y AJUSTAR │     │ Y ENTRENAR    │
 └──────────────┘     └──────────────┘     └───────────────┘
```

1. **Recolectar:** importar CSV, Excel, bases de datos o APIs.
2. **Preparar:** manejar valores faltantes, codificar categorías, escalar variables.
3. **EDA** (análisis exploratorio): visualizar distribuciones, correlaciones y valores atípicos.
4. **Modelar:** dividir en entrenamiento y prueba, ajustar el modelo, generar predicciones.
5. **Evaluar:** RMSE, MAE (regresión) · Accuracy, AUC (clasificación).
6. **Desplegar:** exportar el modelo, construir una app, programar reportes.

### Un ejemplo mínimo de principio a fin

Usaremos el clásico dataset `mtcars`: consumo de combustible de 32 autos. Queremos **predecir las millas por galón (`mpg`)** a partir del peso, los caballos de fuerza y los cilindros.

**Paso 1 — Cargar los datos** (aquí leemos un CSV directamente desde internet):

````python
URL_DATOS = "https://raw.githubusercontent.com/leslysandra/primeros-pasos-python/main/data/"
mtcars = pd.read_csv(URL_DATOS + "mtcars.csv", index_col="modelo")

mtcars.head()
````

````python
mtcars["mpg"].describe()   # variable objetivo: millas por galón
````

**Paso 2 — Explorar:** una primera mirada a la relación entre el peso del auto y su consumo.

````python
plt.figure(figsize=(6, 4))
plt.scatter(mtcars["wt"], mtcars["mpg"], color="darkblue")
plt.title("Peso del auto vs. Consumo (millas por galón)")
plt.xlabel("Peso (miles de libras)")
plt.ylabel("Millas por galón")
plt.show()
````

**Paso 3 — Dividir en entrenamiento (80 %) y prueba (20 %).**
El modelo *aprende* con los datos de entrenamiento y lo *evaluamos* con datos que nunca vio. `random_state=42` fija la "semilla" para que la división aleatoria sea siempre la misma.

````python
from sklearn.model_selection import train_test_split

train, test = train_test_split(mtcars, test_size=0.2, random_state=42)
print("Autos para entrenar:", len(train))
print("Autos para probar:  ", len(test))
````

**Paso 4 — Ajustar un modelo de regresión lineal.**
La fórmula `"mpg ~ wt + hp + cyl"` se lee: *"mpg en función de peso, caballos de fuerza y cilindros"*.

````python
import statsmodels.formula.api as smf

modelo = smf.ols("mpg ~ wt + hp + cyl", data=train).fit()
print(modelo.summary())   # coeficientes, R², p-valores
````

**Paso 5 — Predecir sobre el conjunto de prueba y Paso 6 — Evaluar** con el **RMSE** (raíz del error cuadrático medio): el error típico del modelo, en las mismas unidades que `mpg`.

````python
predicciones = modelo.predict(test)
print("Predicción para el primer auto de prueba:", round(predicciones.iloc[0], 2), "mpg")
print("Valor real:                              ", test["mpg"].iloc[0], "mpg")

rmse = np.sqrt(np.mean((test["mpg"] - predicciones) ** 2))
print("RMSE en el conjunto de prueba:", round(rmse, 3), "mpg")
````

Este es el **esqueleto de todo problema de regresión** que veremos. Clasificación y otros modelos siguen la misma lógica: cargar → dividir → ajustar → predecir → evaluar.

### ✏️ Tu turno — Evaluar un modelo

````python
# TODO: calcula el MAE (error absoluto medio) del modelo en el conjunto de prueba
# Pista: MAE = promedio del valor ABSOLUTO de (real - predicción) → np.abs(...)
mae = np.mean(np.abs(test["mpg"] - predicciones))

print("MAE:", round(mae, 3), "mpg")
revisar(abs(mae - np.mean(np.abs(test["mpg"] - modelo.predict(test)))) < 1e-9)
````
